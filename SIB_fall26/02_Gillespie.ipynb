## Going to Run Birth-Date Process

```julia
using Random
using Plots
```

```julia
### Initial Conditions
t0 = 0.0
N0 = 5
b = 1
d = 0.9
sum_k = b + d
```

```julia
function single_trajectory(t0,N0, b , d , t_max)
    t_max = t_max
    t = t0
    N = N0
    t_arr = [t]
    N_arr = [N]
    while t<t_max && N>0
        B = b*N
        D = d*N
        sum_rate = B+D
        R1 = rand()
        tau = -log(R1)/sum_rate
        t += tau
        R2 = rand()
        if R2 < B/sum_rate
            N += 1
        else
            N -= 1
        end
        if t > t_max 
            break
        end
        push!(t_arr,t)
        push!(N_arr,N)
    end
    return t_arr , N_arr
end
```

```julia
M = 5 ## Ensembles number
t_max = 100
```

```julia
t_ens = []
N_ens = []
for i in 1:M
    t_arr , N_arr = single_trajectory(t0, N0 , b ,d , t_max)
    push!(t_ens,t_arr)
    push!(N_ens,N_arr)
end

```

```julia
p = plot()

for i in 1:M
    plot!(p, t_ens[i], N_ens[i])
end

display(p)
```
