## The task is to Sample from a distribution using basic law of sampling

```julia
using Random
using Plots
```

```julia
x = rand(10^4);
```

```julia
function sampling_disribution(x)
    k = 0.5
    return -log(1-x)/k
end

function given_distribution(x)
    k = 0.5
    return k*exp(-k*x)
end
```

```julia
sample = sampling_disribution.(x);
y = 0: 0.1 :15
actual = given_distribution.(y);
```

```julia
histogram(x, sample, normalize=:pdf)
plot!(y,actual)
```

```julia
mean = sum(sample)/10^5
```

```julia
var = sum(( sample .- mean).^2)/10^5 
```
